```python
# TODO: Combine both functions -> shouldn't be that much effort

def prepare_faces_yawning_detection(new_size = (145, 145)):
    """Returns a list of tuples. The tupley contain the final image-array and the label."""

    folder_training_data = r"../Data/dataset_yawn_eye/train/"
    file_haar = r"../Data/dataset_yawn_eye/haarcascade/haarcascade_frontalface_default.xml"
    labels =["yawn", "no_yawn"]
    result = []

    for label in labels:
        path = os.path.join(folder_training_data, label)
        
        # iterate over all images for the current category
        for img_path in os.listdir(path):
            img_matrix = cv2.imread(img_path)
            face_haar = cv2.CascadeClassifier(file_haar)
            faces_found = face_haar.detectMultiScale(img_matrix, 1.3, 5)

            for (x, y, w, h) in faces_found:
                img = cv2.rectangle(img_matrix, (x, y, w, h), (0, 255, 0), 2)
                # get region of interest
                roi = img[y : y + h, x : x + w]
                resized = cv2.resize(roi, new_size)
                # result-format: [img, label]
                result.append((resized, labels.index(label)))

    return result


def prepare_eyes_blink_detection(new_size = (145, 145)):
    """Returns a list of tuples. The tupley contain the final image-array and the label."""

    folder_training_data = r"../Data/dataset_yawn_eye/train/"
    file_haar = r"../Data/dataset_yawn_eye/haarcascade/haarcascade_frontalface_default.xml"
    labels =["Closed", "Open"]
    result = []

    for label in labels:
        path = os.path.join(folder_training_data, label)

        for img_path in os.listdir(path):
            try:
                img_array = cv2.imread(img_path)
                resized = cv2.resize(img_array, new_size)
                result.append((resized, labels.index(label)))

            except Exception as e:
                print(e)
            
    return result
```
