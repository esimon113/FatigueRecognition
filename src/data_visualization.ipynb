```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
plt.figure()

fig, ax = plt.subplots(2, 2)
fig.set_size_inches(7, 7)


img_yawn = mpimg.imread("../Data/dataset_yawn_eye/train/yawn/106.jpg")
ax[0, 0].imshow(img_yawn)
ax[0, 0].set_title("Gähnen")
ax[0, 0].axis("off")
img_no_yawn = mpimg.imread("../Data/dataset_yawn_eye/train/no_yawn/1029.jpg")
ax[0, 1].imshow(img_no_yawn)
ax[0, 1].set_title("Kein Gähnen")
ax[0, 1].axis("off")
img_open = mpimg.imread("../Data/dataset_yawn_eye/train/Open/_106.jpg")
ax[1, 0].imshow(img_open)
ax[1, 0].set_title("Auge offen")
ax[1, 0].axis("off")
img_closed = mpimg.imread("../Data/dataset_yawn_eye/train/Closed/_103.jpg")
ax[1, 1].imshow(img_closed)
ax[1, 1].set_title("Auge geschlossen")
ax[1, 1].axis("off")

plt.savefig("../imgs/other/sample_data_images.svg")
```
